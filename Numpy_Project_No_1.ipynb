```python
import numpy as np
```

```python
marks = np.array([
    [80,90,75],
    [65,84,72],
    [60,75,90],
    [50,65,89],
    [70,82,91]
])
```

```python
total = np.sum(marks,axis = 1)
print(total)
```

```python
average = np.mean(marks,axis = 1)
print(average)
```

```python
highest = np.max(marks)
print(highest)
```

```python
lowest = np.mean(marks)
print(lowest)
```

```python
sub_avg = np.mean(marks,axis = 0)
print(sub_avg)
```

```python
result = np.where(average >= 50, 'pass', 'fail')
print(result)
```

```python
print("Total Marks",total)
print("Highest marks",highest)
print("Lowest Marks",lowest)
print("Average Marks",average)
print("Sub_Avg",sub_avg)
print("Result",result)
```
